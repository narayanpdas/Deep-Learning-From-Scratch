```python
from model.tokenizer.tokenizer import Tokenizer
from datasets import load_dataset

MAX_TOKEN_LEN = 256
VOCAB_SIZE = 30000
HIDDEN_SIZE = 768
HEAD_DIM = 64
VOCAB_PATH = "model/tokenizer/vocab.txt"
tokenizer = Tokenizer(path=VOCAB_PATH,max_len=MAX_TOKEN_LEN)
```

```python
import pandas as pd
dataset = load_dataset("wikitext", "wikitext-2-raw-v1")
df = dataset['train'].to_pandas()
df_filtered = df[df['text']!=""]
df_ = df_filtered[~df['text'].str.contains("=",na=False)]
df_1k = df_[:1000]
```

```python
import re
non_ascii_regex = re.compile(r'[^\x00-\x7F]+')
def clean_sentence(sentence: str) -> str:
    """Removes non-ASCII characters."""
    return non_ascii_regex.sub('', sentence)
values:dict = {
    "raw_text":[],
    "input_ids":[],
    "token_type_ids":[],
    "attention_mask":[]
}
for _,row in df_1k.iterrows():
    cleaned_text = clean_sentence(row['text'])
    tokenizer.tokenize([cleaned_text])
    values["raw_text"].append(row['text'])
    values["input_ids"].append(tokenizer.input_ids_nd)
    values["token_type_ids"].append(tokenizer.token_type_ids_nd)
    values["attention_mask"].append(tokenizer.attention_mask_nd)
    tokenizer.clear()
```

```python
tokenized_output = pd.DataFrame(values)
```
